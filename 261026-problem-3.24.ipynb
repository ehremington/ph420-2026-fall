```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy.special as sp
import math as math
```

```python
def factorial(x):
    return(sp.factorial(x, exact=True))

def multi_Ein(N, q):
    return(factorial(q+N-1)/(factorial(q)*factorial(N-1)))

def logArray(array):
    return([math.log(x) for x in array])
```

```python
df = pd.DataFrame({'q': range(101)})

N = 50     # number of particles

df['multiplicity'] = multi_Ein(N, df['q'])

df['S/k'] = logArray(df['multiplicity'])

df['kT/epsilon'] = (np.gradient(df['S/k'])/np.gradient(df['q']))**-1

df['C/Nk'] = np.gradient(df['q'])/np.gradient(df['kT/epsilon'])/N
```

```python
df
```

```python
fig0, ax0 = plt.subplots()

ax0.plot(df['q'], df['S/k'], 'o', mfc='None')
```

```python
fig1, ax1 = plt.subplots()

ax1.plot(df['kT/epsilon'], df['C/Nk'], 'ro', mfc='None')
```
