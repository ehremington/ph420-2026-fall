## define some helpful functions

```python
def factorial(x):
    return sp.factorial(x, exact=True)

def multiplicity(N,q):
    return factorial(q+N-1)/(factorial(q)*factorial(N-1))

def logArray(array):
    return [math.log(x) for x in array]
```

```python
table0 = pd.DataFrame()
table0['q_A'] = range(0,100+1, 1)
table0['q_B'] = range(100, 0-1, -1)
table0['multi_A'] = [multiplicity(300,i) for i in table0['q_A']]
table0['multi_B'] = [multiplicity(200,i) for i in table0['q_B']]
table0['multi_total'] = table0['multi_A']*table0['multi_B']
table0['probability'] = table0['multi_total']/table0['multi_total'].sum()
```

```python
fig0, ax0 = plt.subplots()

ax0.bar(table0['q_A'], table0['multi_total'])
ax0.set_xlabel('q_A')
ax0.set_ylabel('total multiplicity')
```
